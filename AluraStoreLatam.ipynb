### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
facturacion_tienda1 = tienda["Precio"].sum()
facturacion_tienda2 = tienda2["Precio"].sum()
facturacion_tienda3 = tienda3["Precio"].sum()
facturacion_tienda4 = tienda4["Precio"].sum()

print("Facturación Tienda 1:", facturacion_tienda1)
print("Facturación Tienda 2:", facturacion_tienda2)
print("Facturación Tienda 3:", facturacion_tienda3)
print("Facturación Tienda 4:", facturacion_tienda4)
```

La tienda con mayor facturación fue la tienda 1 con 1150880400.0 pesos

# 2. Ventas por categoría

```python
categoria_tienda1 = tienda["Categoría del Producto"].value_counts()
categoria_tienda2 = tienda2["Categoría del Producto"].value_counts()
categoria_tienda3 = tienda3["Categoría del Producto"].value_counts()
categoria_tienda4 = tienda4["Categoría del Producto"].value_counts()

print("Tienda 1")
print(categoria_tienda1)

print("\nTienda 2")
print(categoria_tienda2)

print("\nTienda 3")
print(categoria_tienda3)

print("\nTienda 4")
print(categoria_tienda4)
```

La categoría más vendida en todas las tiendas son los muebles.

# 3. Calificación promedio de la tienda


```python
calificacion1 = tienda["Calificación"].mean()
calificacion2 = tienda2["Calificación"].mean()
calificacion3 = tienda3["Calificación"].mean()
calificacion4 = tienda4["Calificación"].mean()

print("Calificación promedio Tienda 1:", calificacion1)
print("Calificación promedio Tienda 2:", calificacion2)
print("Calificación promedio Tienda 3:", calificacion3)
print("Calificación promedio Tienda 4:", calificacion4)
```

La tienda con mejor calificación promedio es la tienda 3 con una clasificación de 4.04

# 4. Productos más y menos vendidos

```python
mas_vendidos1 = tienda["Producto"].value_counts().idxmax()
menos_vendidos1 = tienda["Producto"].value_counts().idxmin()

mas_vendidos2 = tienda2["Producto"].value_counts().idxmax()
menos_vendidos2 = tienda2["Producto"].value_counts().idxmin()

mas_vendidos3 = tienda3["Producto"].value_counts().idxmax()
menos_vendidos3 = tienda3["Producto"].value_counts().idxmin()

mas_vendidos4 = tienda4["Producto"].value_counts().idxmax()
menos_vendidos4 = tienda4["Producto"].value_counts().idxmin()

print("Tienda 1 - Más vendido:", mas_vendidos1)
print("Tienda 1 - Menos vendido:", menos_vendidos1)

print("Tienda 2 - Más vendido:", mas_vendidos2)
print("Tienda 2 - Menos vendido:", menos_vendidos2)

print("Tienda 3 - Más vendido:", mas_vendidos3)
print("Tienda 3 - Menos vendido:", menos_vendidos3)

print("Tienda 4 - Más vendido:", mas_vendidos4)
print("Tienda 4 - Menos vendido:", menos_vendidos4)
```

Los producto más vendidos y menos vendidos varian según la tienda, así que no puedo afirmar que exista una tendencia.

# 5. Envío promedio por tienda

```python
envio1 = tienda["Costo de envío"].mean()
envio2 = tienda2["Costo de envío"].mean()
envio3 = tienda3["Costo de envío"].mean()
envio4 = tienda4["Costo de envío"].mean()

print("Costo de envío promedio Tienda 1:", envio1)
print("Costo de envío promedio Tienda 2:", envio2)
print("Costo de envío promedio Tienda 3:", envio3)
print("Costo de envío promedio Tienda 4:", envio4)
```

El costo de envío promedio más bajo esta en la tienda 4, con un costo de envío promedio de 23459.45

Visualización de datos

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
import pandas as pd

tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

ingresos_totales = [
    facturacion_tienda1,
    facturacion_tienda2,
    facturacion_tienda3,
    facturacion_tienda4
]

df_ingresos = pd.DataFrame({
    'Tienda': tiendas,
    'Ingresos Totales': ingresos_totales
})

plt.figure(figsize=(12,6))

plt.bar(
    df_ingresos['Tienda'],
    df_ingresos['Ingresos Totales'],
    color=['skyblue','salmon','lightgreen','gold']
)

plt.title('Facturación Total por Tienda', fontsize=14, fontweight='bold')
plt.xlabel('Tienda')
plt.ylabel('Ingresos Totales')


def format_millions(x, pos):
    return f'{x/1000000:,.0f}M'

formatter = FuncFormatter(format_millions)
plt.gca().yaxis.set_major_formatter(formatter)


for i, v in enumerate(df_ingresos['Ingresos Totales']):
    plt.text(i, v + 50000000, f'{v:,.0f}', ha='center')


plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
todas_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4])

ventas_totales = todas_tiendas["Categoría del Producto"].value_counts()
ventas_totales_ordenadas = ventas_totales.sort_values(ascending=True)

plt.figure(figsize=(14,6))

plt.barh(
    ventas_totales_ordenadas.index,
    ventas_totales_ordenadas.values,
    color='teal'
)

plt.gca().invert_yaxis()

plt.title('Ventas Totales por Categoría de Producto', fontsize=16, fontweight='bold')
plt.xlabel('Cantidad de Ventas')
plt.ylabel('Categoría del Producto')

for index, value in enumerate(ventas_totales_ordenadas.values):
    plt.text(value, index, f' {value}', va='center')

plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
